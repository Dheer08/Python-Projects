```python
from langchain.agents import create_agent
# This is the Basic GenAI Application
agent0 = create_agent(
    model = "gpt-5",
    tools = [],
    system_prompt = "You are a helpful assistant.",
)
agent0
```

```python
from langchain.agents import create_agent
# This is the Basic GenAI Application

def get_weather_info(city:str) -> str:
    """ 
    Tool Function to get weather information for a given city
    """ 
    return f"The current weather in {city} is sunny with a temperature of 25°C."

agent1 = create_agent(
    model = "gpt-5",
    tools = [get_weather_info],
    system_prompt = "You are a helpful assistant.",
)
agent1
```

```python
agent0.invoke({"messages": [{"role": "user", "content": "What is the weather in New York?"}]})
```

```python
agent1.invoke({"messages": [{"role": "user", "content": "What is the weather in New York?"}]})
# LLM call the tool function and get the weather info based on the doc string provided in the tool function.
```

```python
# The LLM can tell the message it is from the user
response = agent1.invoke({"messages" : "what is the weather in New York?"})
response["messages"]
```
