```python
sample_text = {
    "../Data/text_files/rag_intro.txt": """
    This is the content of sample file 1. Learn RAG."
    RAG stands for Retrieval-Augmented Generation, a technique that combines retrieval of relevant documents with 
    generative models to produce more accurate and contextually relevant responses.RAG is particularly useful in scenarios 
    where the model needs to answer questions or generate text based on specific information that may not be present in its 
    training data. RAG models can retrieve relevant documents from a large corpus and use that information to
     generate more informed and accurate responses.
    """,
     "../Data/text_files/python_intro.txt": """
        This is the content of sample file 1. Learn Python."
        Python is a high-level, interpreted programming language known for its simplicity and readability. It supports multiple
        programming paradigms, including procedural, object-oriented, and functional programming. Python is widely used in
        web development, data analysis, artificial intelligence, scientific computing, and more. Its extensive standard
        library and active community contribute to its popularity, making it a versatile choice for developers of all skill levels.
        """
}      

for file_path, content in sample_text.items():
    with open(file_path, "w",encoding="utf-8") as f:
        f.write(content)

print("✅ Sample text files created successfully.")    
```

```python
# Text Loader
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../Data/text_files/rag_intro.txt",encoding="utf-8")
document = loader.load()
print(document)
```

```python
# Directory Loader
from langchain_community.document_loaders import DirectoryLoader

# Load all text files from the directory
directory_loader = DirectoryLoader(
    "../Data/text_files", 
    glob="*.txt",  # Pattern to match text files
    loader_cls=TextLoader, # Loader class to use
    loader_kwargs={"encoding": "utf-8"}
    )
documents = directory_loader.load()
print(list(documents))
```

```python
# Directory Loader for PDF files
from langchain_community.document_loaders import PyPDFLoader,PyMuPDFLoader,DirectoryLoader

# Load all text files from the directory
directory_loader = DirectoryLoader(
    "../Data/pdf_files", 
    glob="*.pdf",  # Pattern to match pdf files
    loader_cls=PyPDFLoader, # Loader class to use
    )
pdf_documents = directory_loader.load()
pdf_documents
```
