### Messages

Messages are the fundamental unit of context for models in langchain. They represent the input and output of models, carrying both the content and metadata needed to represent the state of conversation when interacting with a LLM. Messages are a object that contains : 
1. Role - Identifies the message type (E.g. system, user)
2. Content - Represents the actual content of the message (like text, images, audio, documents, etc..)
3. Metadata - Optional fields such as response information, message IDs, and token usage

Langchain provides a standard message type that works across all model providers, ensuring consistent behaviour regardless of the model being called

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")

from langchain.chat_models import init_chat_model
model= init_chat_model("openai:gpt-4.1")
```

```python
model.invoke("What is the machine learning ?") # Human Message
# The output of the model is the ai message
```

### Message Prompts

Alternatively, you can pass in a list of messages to the model by providing a list of messages objects
Message Types
1. System Message - Tells the model how to behave and provide context for interaction
2. Human Message - Represent the user input and interactions with the model
3. AI Message - Responses generated by the model, including text content, tool calls and metadata
4. Tool Message - Represent the ouputs of tool calls

### Human Message

A HumanMessage represents user input and interactions. They can contain text, images, audio, files and any other amount of multimodal content

#### System message

A System Message represent an inital set of instructions that primes the mdoel's behaviour. You can use a system message to set the tone, define the model's role and establish guidelines for responses

#### AI Message

An AIMessage represents the output of a model invocation, THey can include multimodal data, tool calls and provider-specific metadata that you can later access

### Tool Message

For models that support tool calling, AI Messages can contain tool calls, Tool messages are used to pass the results of a single tool execution back to the model

```python
from langchain.messages import HumanMessage, AIMessage, SystemMessage

messages = [
    SystemMessage(content="You are a poetry Expert."),
    HumanMessage(content="Write an poem about machine learning ?"),
]
response = model.invoke(messages) 
print(response.content)
```

```python
system_message = SystemMessage(content="You are a helpful assistant.")
messages = [
    system_message,
    HumanMessage(content="How do I create a REST API?"),
]
response = model.invoke(messages)   
print(response.content)
```

```python
## Detailed INfo to the LLM through the system Message you will get a more proper information1
system_message = SystemMessage("""
you are a senior python developer with expertise in web frameworks. Always provide code examples and explain your reasoning.
Be consise but thorough in your explanation.
""")
messages = [
    system_message,
    HumanMessage(content="How do I create a REST API?"),
]
response = model.invoke(messages)   
print(response.content)
```

```python
human_message = HumanMessage(
    content="Hello",
    name="Rakshith", # Optional : Identify the user
    id="msg_123", # Optional : Unique identifier for the message
    )
response = model.invoke([human_message])
print(response.content) 
```

```python
from langchain.messages import HumanMessage, AIMessage, SystemMessage

# Create a AI message Manually
ai_message = AIMessage("I'd be happy to help you with the question")

messages = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="Can you help me?"),
    ai_message, # Insert as if it came from the model, but you can also create your own AIMessage
    HumanMessage(content="Great, What's 2+2?"),
]
response = model.invoke(messages)
print(response.content)
```

```python
# Look at the metadata
print(response.usage_metadata)
```

```python
from langchain.messages import HumanMessage, AIMessage, SystemMessage
from langchain.messages import ToolMessage

# After a model make a tool call
ai_message = AIMessage(
    content = "",
    tool_calls=[{
        "name": "get_weather",
        "args" : {"location": "Dublin", "unit": "celsius"},
        "id" : "Call_123"
    }]
)

weather_result = "Cold, 10 degrees Celsius"
tool_message = ToolMessage(
    content = weather_result,
    tool_call_id = "Call_123" # Must match the tool call id from the AIMessage
)

messages = [
    HumanMessage(content="What is the weather in Dublin?"),
    ai_message, # Model's tool call
    tool_message # Tool's response
]
response = model.invoke(messages)
print(response.content)
print(response.usage_metadata)
print("Tool Message : ",tool_message.content)
```
